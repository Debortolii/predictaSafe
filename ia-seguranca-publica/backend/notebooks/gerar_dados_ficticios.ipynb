```python
import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta

# Inicializar Faker
fake = Faker('pt_BR')

# Definir número de registros
num_registros = 1000

# Listas de dados para simulação
tipos_desordem = ['Vandalismo', 'Pichação', 'Barulho Excessivo', 'Lixo Acumulado', 'Aglomeração Suspeita', 'Outros']
locais_desordem = ['Praça Central', 'Estação de Metrô', 'Área Comercial', 'Bairro Residencial A', 'Bairro Residencial B', 'Parque Urbano']
periodos_dia = ['Manhã', 'Tarde', 'Noite', 'Madrugada']
status_ocorrencia = ['Reportada', 'Em Análise', 'Resolvida', 'Cancelada']
niveis_severidade = ['Baixa', 'Média', 'Alta', 'Crítica']

# Gerar dados fictícios
data = []
for _ in range(num_registros):
    data_ocorrencia = fake.date_time_between(start_date='-1y', end_date='now')

    # Simular coordenadas geográficas próximas ao Brasil (exemplo, ajuste para sua cidade se quiser)
    latitude = random.uniform(-23.0, -24.0) # Exemplo para São Paulo, ajuste conforme necessário
    longitude = random.uniform(-46.0, -47.0) # Exemplo para São Paulo, ajuste conforme necessário

    registro = {
        'ID_Ocorrencia': fake.uuid4(),
        'Data_Hora': data_ocorrencia,
        'Tipo_Desordem': random.choice(tipos_desordem),
        'Local_Desordem': random.choice(locais_desordem),
        'Latitude': latitude,
        'Longitude': longitude,
        'Periodo_Dia': random.choice(periodos_dia),
        'Nivel_Severidade': random.choice(niveis_severidade),
        'Status_Ocorrencia': random.choice(status_ocorrencia),
        'Observacoes': fake.sentence(),
        'Risco_Previsao': random.choices([0, 1], weights=[0.7, 0.3], k=1)[0] # 0 para baixo risco, 1 para alto risco
    }
    data.append(registro)

df_desordens = pd.DataFrame(data)

# Salvar o DataFrame em um arquivo CSV
# O CSV será salvo na mesma pasta onde o notebook está.
df_desordens.to_csv('desordens_ficticias.csv', index=False)

print(f"Dados fictícios gerados e salvos em 'desordens_ficticias.csv' com {num_registros} registros.")
print(df_desordens.head())
```
